### Regression Analysis:

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
var
```

```python
var.isnull().sum()
```

```python
var['Car_Name'] = var['Car_Name'].astype('category')
var['Car_Name'] = var['Car_Name'].cat.codes

var['Fuel'] = var['Fuel'].astype('category')
var['Fuel'] = var['Fuel'].cat.codes

var['Seller_Type'] = var['Seller_Type'].astype('category')
var['Seller_Type'] = var['Seller_Type'].cat.codes

var['Transmission'] = var['Transmission'].astype('category')
var['Transmission'] = var['Transmission'].cat.codes

var['Owner'] = var['Owner'].astype('category')
var['Owner'] = var['Owner'].cat.codes
```

```python
x=var.drop(columns="Selling_Price")
x
```

```python
y=var["Selling_Price"]
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.25, random_state=0)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
c=model.intercept_
c
```

```python
m=model.coef_
m
```

```python
predicted_y_test=model.predict(x_test)
predicted_y_test
```

```python
mse=mean_squared_error(y_test, predicted_y_test)
mse
```

```python
r2=r2_score(y_test, predicted_y_test)
r2
```

```python
print(f" R^2 Value is: {r2*100:.2f}%")
```

#### Insight: In this modelling scenario, we are getting the R-squared value of 67.82%, it means that 67.82% of times, the independent variables would be able to explain the changes in the depenedent variable.
